```python
def portfolio_performance(weights):

    # Annualized portfolio return
    portfolio_return = np.dot(
        weights,
        mean_daily_returns
    ) * 252

    # Portfolio variance
    portfolio_variance = np.dot(
        weights,
        np.dot(covariance_matrix, weights)
    )

    # Annualized portfolio volatility
    portfolio_volatility = (
        np.sqrt(portfolio_variance) * np.sqrt(252)
    )

    # Sharpe ratio
    sharpe_ratio = (
        portfolio_return - risk_free_rate
    ) / portfolio_volatility

    return portfolio_return, portfolio_volatility, sharpe_ratio
```

```python
from scipy.optimize import minimize
```

```python
def negative_sharpe(weights):

    portfolio_return, portfolio_volatility, sharpe_ratio = (
        portfolio_performance(weights)
    )

    return -sharpe_ratio
```

```python
constraints = {
    "type": "eq",
    "fun": lambda weights: np.sum(weights) - 1
}

bounds = tuple(
    (0, 1) for _ in range(5)
)
```

```python
initial_weights = np.array([
    0.20,
    0.20,
    0.20,
    0.20,
    0.20
])
```

```python
optimization = minimize(
    negative_sharpe,
    initial_weights,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints
)
```

```python
optimized_weights = optimization.x

print("Optimized Weights:")
print("ITC =", optimized_weights[0])
print("L&T =", optimized_weights[1])
print("Adani Ports =", optimized_weights[2])
print("Wipro =", optimized_weights[3])
print("Indian Oil =", optimized_weights[4])
```

```python
# Calculate portfolio risk contribution

def risk_contribution(weights, cov_matrix):
    
    portfolio_variance = np.dot(weights.T, np.dot(cov_matrix, weights))
    
    marginal_contribution = np.dot(cov_matrix, weights)
    
    contribution = weights * marginal_contribution
    
    percentage_contribution = contribution / portfolio_variance
    
    return percentage_contribution
```

```python
optimized_risk_contribution = risk_contribution(
    optimized_weights,
    covariance_matrix
)

risk_contribution_df = pd.DataFrame({
    "Stock": stocks.keys(),
    "Weight": optimized_weights * 100,
    "Risk Contribution": optimized_risk_contribution * 100
})

risk_contribution_df = risk_contribution_df.round(2)

print(risk_contribution_df)
```
